```python
import pandas as pd 
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('../data/Titanic_processed.csv') #Stored the processed data in df
```

```python
features = df.drop(columns=['Survived'])
labels = df['Survived']
```

```python
X_train, X_test, y_train, y_test = train_test_split(features, labels, test_size=0.3, random_state=42)
```

```python
# -------------------------------------------------------------------------- #

# model = LogisticRegression(max_iter= 1000) #Initializing the model
# model.fit(X_train, y_train) #Applying the fit method
# predictions = model.predict(X_test)

# -------------------------------------------------------------------------- #

# Implementing Standard Scaler to improve accuracy

from sklearn.preprocessing import StandardScaler

scalar = StandardScaler()

X_train_scaled = scalar.fit_transform(X_train)
X_test_scaled = scalar.transform(X_test)

model = LogisticRegression()
model.fit(X_train_scaled, y_train)
predictions = model.predict(X_test_scaled)
```

```python
from sklearn.metrics import confusion_matrix

c_matrix = confusion_matrix(y_test, predictions)
print(c_matrix, end = '\n\n')

from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, predictions)
print(f"Accuracy: {accuracy * 100:.2f}%", end = '\n\n')

print(predictions)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, predictions))
```
